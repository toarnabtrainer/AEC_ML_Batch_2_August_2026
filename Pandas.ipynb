# **Introduction to Pandas**
* **It is often said that 80% of data analysis activity is spent on the data cleaning and preparation of our data. To get the grip over this activity, this section will focus on a small but important aspect of data manipulation and cleaning with Pandas.**

## **Inbuilt Data Structure in Pandas**
**There are two types of data structures are there in Pandas -** <br>

* **Series -** It is one-dimensional labeled array capable of holding any data type (integer, string, float, Python objects etc.) of data. The axis is collectively referred to as index. <br>

* **Data Frame -** It is a two-dimensional labeled data structure with columns of potentially different types. You can think of it like a spreadsheet or SQL Table or a Series of objects <br>

## **Series Data Structure**
**pandas.core.series.Series(data, index, dtype, copy)** <br>
* **data -** data takes various forms like ndarray, list, constants, dictionary etc. <br>
* **index -** it is unique and hashable for easy identification. <br>
* **dtype -** it is for data type. <br>
* **copy -** copy data, and its default value is False. It only affects for Series or one dimensional ndarray inputs. <br>

```python
# creating empty series
s = pd.Series()
print(s, type(s), len(s), id(s))
```

```python
# create series from nd-array
arr_data = np.array([100, 300, 500, 400])
print(arr_data, type(arr_data))
s = pd.Series(data = arr_data)
print(s)
print(s.dtype, type(s), len(s))
```

```python
# create series from nd-array
arr_data = np.array([100, 300, 500, 400])
s = pd.Series(data = arr_data, copy = False)
arr_data[0] = 999
s[1] = 888
print(arr_data, type(arr_data))
print(s)
```

```python
# create series from nd-array
arr_data = np.array([100, 300, 500, 400])
s = pd.Series(data = arr_data, copy = True)
arr_data[0] = 999
s[1] = 888
print(arr_data, type(arr_data))
print(s)
```

```python
arr_data = np.array(['apple', 'pineapple', 'kiwi', 'guava', 'mango'])
s = pd.Series(data = arr_data)
print(s)
print(s[0], s[3])  # indexing
print(s[1:4])  # slicing
```

```python
arr_data = np.array(['apple', 'pineapple', 'kiwi', 'guava', 'mango'])
s = pd.Series(data = arr_data, index = [101, 103, 102, 101, 105])
print(s)
print(s[102])
print(s[101])
```

```python
arr_data = np.array(['apple', 'pineapple', 'kiwi', 'guava', 'mango'])
s = pd.Series(data = arr_data, index = ["fr-1", "fr-2", "fr-3", "fr-4", "fr-5"])
print(s)
print(s['fr-1'], s[0])
print(s['fr-4'], s[3])
```

```python
# create series from dictionary
dict_data = {'apple':100, "banana":300, "lemon":450, "lime":490}
s = pd.Series(data = dict_data)
print (s)
s = pd.Series(data=dict_data, index=['apple','lime','banana','coconut'])
print(s)
```

```python
# create series from a scalar
s = pd.Series(5, index = [2, 1, 4, 3, 0])
print(s)
s = pd.Series(5, index = ['num1', 'num2', 'num3', 'num4'])
print(s)
```

```python
# create series from a list
s = pd.Series(data=[100,200,400,300], index=['red','blue','pink','black'])
print(s)
print(s[0], s['red'])
print(s[1:3])
print(s[-3:-1])
print(s['blue':'black'])
print(s[['blue','pink','black']])
```

```python
# right outer join: matching rows from left and all rows from right
merge_right = pd.merge(users, user4, on='name', how="right")
display(merge_right)
```

```python
# strategy-1: dropping rows containing null values
df = merge_left.copy()
df.dropna(inplace = True, ignore_index = True) # drop row if null in any attribute
display(df)
```

```python
df = merge_left.copy()
# drop row if null in all attributes
df.dropna(how = "all", inplace = True, ignore_index = True)
display(df)
```

```python
# strategy-2: filling up missing values
df = merge_left.copy()
df.fillna(0, inplace = True)
display(df)
```

```python
df = merge_left.copy()
mean_height = np.round(df.height.mean(), 2)
print("Mean height is", mean_height)
df.fillna(mean_height, inplace = True)
display(df)
```

```python
df = merge_left.copy()
median_height = np.round(df.height.median(), 2)
print("Median height is", median_height)
df.fillna(median_height, inplace = True)
display(df)
```

```python
df = merge_left.copy()
display(df)
mode_height = df.height.mode()
print(mode_height)
print(mode_height[0])
df.fillna(mode_height[0], inplace = True)
display(df)
```

```python
# forward fill
df = merge_left.copy()
df.sort_values(by=['gender','job'], ascending=[True,False], inplace=True, ignore_index=True)
display(df)
# df.fillna(method = 'pad', inplace = True)
df.ffill(inplace = True)
display(df)
```

```python
# backward fill
df = merge_left.copy()
df.sort_values(by=['gender','job'], ascending=[True,False], inplace=True)
print(df.index)
df.index = range(0, len(df))
print(df.index)
display(df)
# df.fillna(method = 'bfill', inplace = True)
df.bfill(inplace = True)
display(df)
```

```python
# depending on gender, fill up null heights with gender specific average heights
df = merge_left.copy()
display(df)
print("Male average height:", df[df.gender == 'M'].height.mean())
print("Female average height:", df[df.gender == 'F'].height.mean())
df['height'] = df.height.fillna(df.groupby('gender')['height'].transform('mean'))
display(df)
```
